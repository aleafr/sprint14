## Objetivo do estudo

O objetivo deste projeto é analisar os dados disponíveis para responder a questões relacionadas com a publicação de livros, avaliações dos utilizadores e desempenho de autores e editoras.

### 1. Quantos livros foram lançados depois de 1 de janeiro de 2000?

````python
query = """
SELECT COUNT(*) AS total_books_after_2000
FROM books
WHERE publication_date > '2000-01-01'
"""
pd.read_sql(query, con=engine)
````

````python
query = """
SELECT COUNT(*) AS total_books
FROM books
"""
pd.read_sql(query, con=engine)
````

Foram identificados 819 livros publicados após 1 de janeiro de 2000.

Ao comparar com o número total de livros na base de dados, é possível avaliar a proporção de obras recentes. Isso ajuda a perceber se o catálogo está mais focado em livros contemporâneos ou inclui uma grande quantidade de obras mais antigas.

A presença de um número significativo de livros publicados após 2000 pode indicar que a base de dados está relativamente atualizada e alinhada com tendências modernas de leitura.

### 2. Número de avaliações e classificação média por livro

````python
query = """
SELECT
    b.book_id,
    b.title,
    COUNT(r.rating) AS total_ratings,
    AVG(r.rating) AS avg_rating
FROM books b
LEFT JOIN ratings r ON b.book_id = r.book_id
GROUP BY b.book_id, b.title
ORDER BY total_ratings DESC
"""
pd.read_sql(query, con=engine)
````

A análise mostra o número total de avaliações e a classificação média de cada livro.

Observa-se que alguns livros possuem um número elevado de avaliações, o que pode indicar maior popularidade ou visibilidade entre os utilizadores. Por outro lado, existem livros com poucas ou nenhuma avaliação, o que pode afetar a fiabilidade da sua classificação média.

A classificação média permite identificar a perceção geral dos leitores sobre cada livro. No entanto, é importante considerar que livros com poucas avaliações podem apresentar médias pouco representativas.

Assim, para uma análise mais robusta, seria recomendável considerar também o volume de avaliações ao interpretar a qualidade de cada livro.

### 3. Editora com mais livros com mais de 50 páginas

````python
query = """
SELECT 
    p.publisher,
    COUNT(b.book_id) AS total_books
FROM books b
JOIN publishers p ON b.publisher_id = p.publisher_id
WHERE b.num_pages > 50
GROUP BY p.publisher
ORDER BY total_books DESC
LIMIT 1
"""
pd.read_sql(query, con=engine)
````

A análise identifica a editora que publicou o maior número de livros com mais de 50 páginas.

Este filtro permite excluir publicações muito curtas, focando apenas em livros mais completos. A editora apresentada no resultado é, assim, a que possui maior volume de publicações relevantes dentro deste critério.

### 4. Autor com maior média de classificação (mínimo 50 avaliações)

````python
query = """
WITH book_ratings AS (
    SELECT
        b.book_id,
        b.author_id,
        COUNT(r.rating_id) AS num_ratings,
        AVG(r.rating) AS avg_rating
    FROM books AS b
    JOIN ratings AS r ON b.book_id = r.book_id
    GROUP BY b.book_id, b.author_id
    HAVING COUNT(r.rating_id) >= 50
),

author_avg_ratings AS (
    SELECT
        br.author_id,
        a.author,
        AVG(br.avg_rating) AS avg_author_rating
    FROM book_ratings AS br
    JOIN authors AS a ON br.author_id = a.author_id
    GROUP BY br.author_id, a.author
)

SELECT *
FROM author_avg_ratings
ORDER BY avg_author_rating DESC
LIMIT 1;
"""

pd.read_sql(query, con=engine)
````

### 5. Número médio de avaliações de utilizadores com mais de 50 avaliações

````python
query = """
SELECT AVG(review_count) AS avg_reviews
FROM (
    SELECT
        username,
        COUNT(review_id) AS review_count
    FROM reviews
    WHERE username IN (
        SELECT
            username
        FROM ratings
        GROUP BY username
        HAVING COUNT(rating_id) > 50
    )
    GROUP BY username
) AS review_counts
"""

pd.read_sql(query, con=engine)
````
